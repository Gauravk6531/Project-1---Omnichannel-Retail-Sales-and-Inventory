```python
# 2. Standardize column names to lowercase snake_case
print("Standardizing column names...")
df_sales.columns = df_sales.columns.str.strip().str.lower().str.replace(' ', '_')
```

```python
# 3. Convert date column from DD-MM-YYYY to standard SQL format (YYYY-MM-DD)
print("Reformatting transaction dates...")
df_sales['transaction_date'] = pd.to_datetime(df_sales['transaction_date'], format='%d-%m-%Y').dt.strftime('%Y-%m-%d')
```

```python
print(df_sales.columns.tolist())
# Dropping reason to return because all entries null.
df_sales = df_sales.drop('reason_of_return', axis=1)
```
