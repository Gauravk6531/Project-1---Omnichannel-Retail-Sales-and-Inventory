```python
# start date and end date is in string converting it to date time object
df_inventory['Start Date'] = pd.to_datetime(df_inventory['Start Date'])
df_inventory['End Date'] = pd.to_datetime(df_inventory['End Date'])
df_inventory.info()
```

```python
# ==========================================
# 2. FILTERING THE INVENTORY LEDGER (df_inventory)
# ==========================================
# Convert Start Date to a datetime object
df_inventory['Start Date Parsed'] = pd.to_datetime(df_inventory['Start Date'])

# Method A: Filter for inventory rows that INITIATED/STARTED in June 2025
df_inventory_june2025_start = df_inventory[
    (df_inventory['Start Date Parsed'].dt.year == 2025) & 
    (df_inventory['Start Date Parsed'].dt.month == 6)
].copy()

# Method B (Optional but recommended for deep inventory tracking):
# If you want to find ANY inventory row that was ACTIVE during June 2025
# (even if it started before June or extended past June), use a window overlap:
df_inventory_june2025_active = df_inventory[
    (df_inventory['Start Date'] <= '2025-06-30') & 
    (df_inventory['End Date'] >= '2025-06-01')
].copy()


# ==========================================
# 3. VERIFICATION
# ==========================================
print(f"June 2025 Inventory Rows (by Start Date): {len(df_inventory_june2025_start)}")
print(f"June 2025 Inventory Rows (Active during month): {len(df_inventory_june2025_active)}")


df_inventory_june2025_start
```
